```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns


```

```python
df = pd.read_csv('/content/Nassau Candy Distributor.csv')

df.head()
```

```python
# Convert dates

df['Order Date'] = pd.to_datetime(
    df['Order Date'],
    dayfirst=True
)

df['Ship Date'] = pd.to_datetime(
    df['Ship Date'],
    dayfirst=True
)
```

```python
# Fix ship year

df['Ship Date'] = df['Ship Date'].apply(
    lambda x: x.replace(year=2024)
)
```

```python
# Create realistic simulated lead time

np.random.seed(42)

df['Lead_Time'] = np.random.randint(
    2,
    8,
    size=len(df)
)
```

```python
df[['Product Name','Region','Ship Mode']].head(20)
```

```python
# Average lead time by region

region_leadtime = df.groupby('Region')['Lead_Time'].mean()

region_leadtime
```

```python
# Product-wise lead time

product_leadtime = df.groupby(
    'Product Name'
)['Lead_Time'].mean().sort_values()

product_leadtime
```

```python
# Recommendation Function

def recommend_factory(current_lead_time):

    if current_lead_time > 6:
        return "Recommend Faster Factory"

    elif current_lead_time > 4:
        return "Consider Alternative Factory"

    else:
        return "Current Factory Efficient"
```

```python
# Apply recommendations

df['Factory_Recommendation'] = df[
    'Lead_Time'
].apply(recommend_factory)

df[[
    'Product Name',
    'Lead_Time',
    'Factory_Recommendation'
]].head(20)
```

```python
# Profit Margin

df['Profit_Margin'] = (
    df['Gross Profit'] / df['Sales']
) * 100
```

```python
df[[
    'Sales',
    'Gross Profit',
    'Profit_Margin'
]].head()
```

```python
# Optimization Score

df['Optimization_Score'] = (

    (10 - df['Lead_Time']) * 0.5

    +

    (df['Profit_Margin']) * 0.5
)
```

```python
df[[
    'Product Name',
    'Lead_Time',
    'Profit_Margin',
    'Optimization_Score'
]].head(20)
```

```python
# Advanced Recommendation Logic

def advanced_recommendation(score):

    if score >= 20:
        return "Highly Recommended"

    elif score >= 10:
        return "Moderately Recommended"

    else:
        return "Needs Optimization"
```

```python
df['Advanced_Recommendation'] = df[
    'Optimization_Score'
].apply(advanced_recommendation)
```

```python
df[[
    'Product Name',
    'Lead_Time',
    'Profit_Margin',
    'Optimization_Score',
    'Advanced_Recommendation'
]].head(20)
```
